```python
#Обчислення факторіалу за допомогою циклу for
def factorial_for(n):
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i
    return factorial

#Рекурсивний алгоритм обчислення факторіалу
def fac(n):
    if n == 0:
        return 1
    return fac(n - 1) * n

#Рекурсивний алгоритм обчислення чисел Фібоначчі
def fibonacci(n):
    if n == 0:
        return 0
    if n in (1, 2):
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

#Алгоритм сортування злиттям (Merge Sort)
def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result += left[i:]
    result += right[j:]
    return result

def mergesort(list):
    if len(list) < 2:
        return list
    middle = len(list) // 2
    left = mergesort(list[:middle])
    right = mergesort(list[middle:])
    return merge(left, right)
```

| Алгоритм                          | Часова складність   | Складність по пам'яті |
|-----------------------------------|---------------------|-----------------------|
| Факторіал (циклом for)           | O(n)                | O(1)                  |
| Факторіал (рекурсивно)            | O(n)                | O(n)                  |
| Числа Фібоначчі (рекурсивно)     | O(2^n)              | O(n)                  |
| Сортування злиттям (Merge Sort)  | O(n log n)          | O(n)                  |
